# Théorie sur l'horloge atomique

L'horloge atomique est un système introduisant en 2 composantes: Un oscillateur à quartz (comme une horloge électronique classique) et les atomes de Césium (ou autres matériaux). **Le quartz est calibrer pour vibrer à exactement la même fréquence que les atomes de césium.**

L'horloge est composée d'un quartz ayant des propriétés piezoélectriques. Un courant passe à travers le piézoélectrique et la différence de potentiel le fait vibré à environ 9 GHz sous forme de micro-ondes. C'est ce signal qui sert d'horloge. Or, les propriétés piézoélectriques du quartz varient en fonction de sa température ce qui peut avoir un impact sur la phase du signal émis par le quartz.

Afin de corriger le potentiel déphasage, on utilise du Césium afin de déterminer le déphasage du quartz par rapport à la fréquence naturelle $w_0$ du Césium.

L'atome est originellement dans son état fondamental $|1> = \left(^1_0\right)$. On le fait passé dans un tube où on laisse passé, sur une court temps T, les micro-ondes du quartz telle impulsion $Û_{\frac{\pi}{2}}$. Cette impulsion oriente l'atome dans une superposition d'état. À cet instant, **la phase du césium et du quartz sont parfaitement aligné**
$$\psi(0) = Û_{\frac{\pi}{2}}\left|1\right> = \frac{1}{\sqrt{2}}\left(\left|1\right> + \left|2\right>\right)\\
\psi(0) =\frac{1}{\sqrt{2}}\left(^1_1\right)
$$

En sortant du tube, on le césium **n'est plus sous l'influence du quartz et évolue pendant un temps T avec une fréquence $\omega_0$** selon son propre hamiltonien:
$$
\left|\psi(T)\right> = Û_0(T)\left|\psi(0)\right>\\
\left|\psi(T)\right> = \frac{1}{\sqrt{2}}\left(^{e^{\frac{-iE_1T}{\hbar}}}_{e^{\frac{-iE_2T}{\hbar}}}\right)\\
\left|\psi(T)\right> = \frac{e^{\frac{-iE_1T}{\hbar}}}{\sqrt{2}}\left(^1_{e^{\frac{-i\omega_0T}{\hbar}}}\right)
$$

Par la suite, on fait repasser l'atome dans un tube sous l'influence du champs magnétique du quartz. Si le quartz est encore ajusté à la fréquence du césium, la différence de phase devrait être nulle. Or, s'il y a un décalage entre la période du quartz et celle du césium, on sait qu'il y a un déphasage. Sachant que $\phi_{atome} = \omega_0T$ et $\phi_{quartz} = \omega T$, le déphasage total est $\Delta \phi = (\omega_0 - \omega)T$. Or, pour des raisons de calibration, un système embarquer fait légèrement diverger la fréquence d'oscillation sous laquelle le cesium est soumis à un interval répéter. On utilisera plus tard ces fréquences tel que: $\omega_- = \omega_0 - \Delta\omega_{mod}$ et $\omega_+ = \omega_0 + \Delta\omega_{mod}$


$$
\left|\psi_{final}\right> = Û_{\frac{\pi}{2}}(T)\left|\psi(T)\right>\\
\left|\psi_{final}\right> = \frac{1}{2}\left[\left(1-e^{-i(\omega_0-\omega)T}\right)\left|1\right> + e^{-i\omega T}\left(1+e^{-i(\omega_0-\omega)T}\right)\left|2\right>\right]
$$

Si on cherche la probabilité de mesurer l'état $\left|1\right> $ ou $\left|2\right>$, on a:

$$
P_1 = \left|\left<1\right|\left|\psi_{final}\right>\right|^2\\
P_1 = \left|\left<1\right|\frac{1}{2}\left[\left(1-e^{-i(\omega_0-\omega)T}\right)\left|1\right> + e^{-i\omega T}\left(1+e^{-i(\omega_0-\omega)T}\right)\left|2\right>\right]\right|^2\\
P_1 = \left|\frac{1}{2}\left[\left(1-e^{-i(\omega_0-\omega)T}\right)\right]\right|^2\\
$$

Finalement, on trouve que la probabilité d'obtenir un certain état dépend du déphasage entre le quartz et le césium
$$
P_1 = \frac{1}{2}\left[1-\cos(\Delta \phi)\right]\\
P_2 = \frac{1}{2}\left[1+\cos(\Delta \phi)\right]\\

$$

Puisque l'arcos ne donne pas de précision sur le signe du déphasage, on utilise $\omega_-$ et $\omega_+$ pour trouver le déphasage précis.

$$
\Delta P = P_2(\omega_+) - P_2(\omega_-)\\
\delta \omega = \frac{\Delta P}{T}
$$


La fréquence d'oscillation est donc corrigée plusieurs fois par seconde afin de transmettre le meilleur signal possible. Les erreurs sur la mesure sont: le buir de projection quantique, l'effet Doppler (agitation thermqique), effet Zeeman, rayonnement coprs noir, effets relativistes
